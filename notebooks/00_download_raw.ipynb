```python
import os
import urllib.request

base_url = "https://data.binance.vision/data/spot/monthly/klines/BTCUSDT/1m/"
volume_path = "/Volumes/workspace/vol_project/raw/"

for year in range(2020, 2027):
    for month in range(1, 13):

        # stops after September 2026
        if year == 2026 and month > 9:
            break

        month_str = f"{year}-{month:02d}"
        filename = f"BTCUSDT-1m-{month_str}.zip"

        url = base_url + filename
        dest = volume_path + filename

        if os.path.exists(dest):
            print(f"skipping {month_str}")
            continue

        urllib.request.urlretrieve(url, dest)
        print(f"downloaded {month_str}")
```

```python
import zipfile

csv_path = volume_path + "csv/"
os.makedirs(csv_path, exist_ok=True)

for filename in sorted(os.listdir(volume_path)):
    if not filename.endswith(".zip"):
        continue

    csv_name = filename.replace(".zip", ".csv")
    if os.path.exists(csv_path + csv_name):
        print(f"skipping {csv_name}")
        continue

    with zipfile.ZipFile(volume_path + filename) as z:
        z.extractall(csv_path)
    print(f"extracted {csv_name}")
```
